### resetting index

```python
import pandas as pd

salaries = [4500, 0, 9500, 1200, 3600, 4520]
departments = ["hr", "marketing", "uix designer", "data analyst", "data engineer", "developer"]

salary_series = pd.Series(salaries, index = departments, name = "Salary")
```

```python
salary_series
```

```python
# reset index returns a data frame and changes does not happen in original variable
salary_series.reset_index()
```

```python
salary_series
```

```python
updated_salary = salary_series.reset_index(drop=True)
```
