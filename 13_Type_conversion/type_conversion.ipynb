### Type conversion

```python
import numpy as np
import pandas as pd
```

```python
scores_values = np.array([220, 520, -85, 0, 62, 42, 32, 10])
scores_series = pd.Series(scores_values, name = "Scores")
```

```python
scores_series
```

by using astype() it does not modify the original data type but prints as output


```python
scores_series.astype(float)
```

```python
scores_series.astype(bool)
```

```python
# object and string both same
scores_series.astype(str)
```
