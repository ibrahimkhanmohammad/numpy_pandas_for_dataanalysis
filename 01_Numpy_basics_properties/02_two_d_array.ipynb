```python
import numpy as np

numbers = [[1, 2, 3],
           [4, 6, 8]]
num_array = np.array(numbers)

num_array
```

```python
# ndim - it returns the dimension of the array
num_array.ndim
```

```python
# shape - it returns the shape of the array i.e; len of array
num_array.shape
```

```python
# size - it returns the size of the array (total elements)
num_array.size
```

```python
# dtype - it returns the datatype of elements inside the array (every element should be unique data type)
num_array.dtype
```
