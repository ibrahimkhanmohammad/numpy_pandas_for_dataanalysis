 Make a multiplication table from 1 to 10 without any loop      
Hint: multiply a row vector by a column vector
        

```python
import numpy as np

# Create a row vector
rows = np.arange(1, 11).reshape(1, 10)

# Create a column vector
cols = np.arange(1, 11).reshape(10, 1)

# Multiply using broadcasting
table = rows * cols

print(table)
print("Shape:", table.shape)
```
