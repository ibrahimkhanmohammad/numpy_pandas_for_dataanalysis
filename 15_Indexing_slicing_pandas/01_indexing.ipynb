### Single element indexing


```python
import pandas as pd

salary_series = pd.Series([2500, 4500, 0, 5320, 2100, 7850], name = "Salary")
```

```python
salary_series
```

```python
print(salary_series[0])
print(salary_series[2])
print(salary_series[5])
```

### Multiple element indexing

```python
import pandas as pd

salary_series = pd.Series([2500, 4500, 0, 5320, 2100, 7850], name = "Salary")
```

```python
salary_series
```

```python
print(salary_series[[0, 4]])
print(salary_series[[2,4]])
print(salary_series[[3, 5]])
```
