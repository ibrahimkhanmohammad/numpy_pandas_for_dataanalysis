### Where function

#### where function always gives new array

```python
import numpy as np
```

```python
# convert grades "pass" if score >= 50 else "fail"

scores = np.array([45, 67, 32, 90, 58])
result = np.where(scores >= 50, "pass", "fail")
result
```

```python
# make numbers +ve replace 0 with -ve

nums = np.array([1, -5, 65, 40, -8])
positives = np.where(nums < 0, 0, nums)
positives
```

```python
# apply 10% discount if price > 100

prices = np.array([91, 95, 160, 120, 150, 45])
d_prices = np.where(prices > 100, prices * 0.9, prices)
d_prices
```

```python
# label even if divisible by 2 else odd

nums = np.array([12, 1, 10, 45, 58, 95, 77, 32])
evens = np.where(nums % 2 == 0, "even", "odd")
evens
```

```python
# cap values at a max of 100(upper bound)

nums = np.array([42, 120, 132, 54, 65, 98, 200])
capped = np.where(nums > 100, 100, nums)
capped
```

```python
# label adult if age >= 18 else kid

ages = np.array([8, 18, 23, 21, 16, 14, 48, 36, 3])
adults = np.where(ages >= 18, "adult", "kid")
adults 
```
