### iloc method

Use iloc to select data by position. It ignores the index labels and works with integer locations.

```python
import pandas as pd

salaries = [2500, 0, 9500, 4600, 8500, 3250]
role = ["designer", "engineer", "developer", "doctor", "chef", "mechanic"]

salaries_series = pd.Series(salaries, index = role, name = "Salaries")
role_series = pd.Series(role, name = "Role")
```

```python
salaries_series
```

```python
salaries_series.iloc[0]
```

```python
salaries_series.iloc[-1]
```

```python
salaries_series.iloc[1:5]
```

```python
salaries_series.iloc[[1,3,5]]
```

```python
salaries_series.iloc[::-1]
```
