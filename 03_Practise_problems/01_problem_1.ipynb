1. Create a NumPy array containing the numbers
1 to 10 and arrange it into 5 rows and 2 columns.

```python
import numpy as np

arr = np.arange(1, 11, 1).reshape(5, 2)
print(arr)
```

2. Create another NumPy array of random decimal
values between 0 and 1 in a 3 x 3 shape.

```python
import numpy as np

rng = np.random.default_rng(42)
arr = rng.random((3, 3))
arr
```
