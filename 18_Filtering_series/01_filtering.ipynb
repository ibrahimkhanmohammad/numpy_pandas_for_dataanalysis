### filtering series

```python
import numpy as np
import pandas as pd

salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing", "intern", "sales", "support"])

salary_series = pd.Series(salary, index = department, name = "Salary")
```

```python
salary_series
```

```python
salary_series >= 50000

salary_series.loc[salary_series >= 50000]
```

```python
salary_series.loc[(salary_series >= 50000) & (salary_series <= 60000)]
```

```python
salary_series[salary_series != 0]
```

```python
salary_series.loc[(salary_series >= 50000) | (salary_series <= 60000)]
```

```python
salary_series.loc[~(salary_series >= 50000)]
```

```python
salary_series.loc[salary_series.between(45000, 60000)]
```

```python
mask = salary_series > 50000
print(salary_series[mask])
salary_series[mask].count()
```

better way to count is by using .sum()

.sum()	Counts True values

```python
(salary_series > 50000).sum()
```

any() — At least one condition is True

```python
(salary_series > 50000).any()
```

all() — Every condition must be True

```python
(salary_series > 50000).all()
```

```python
salary_series.loc[salary_series <= 45000] = 50000
salary_series
```
