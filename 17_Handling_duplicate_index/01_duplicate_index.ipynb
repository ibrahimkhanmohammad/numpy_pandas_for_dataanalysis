### duplicate indexes

Duplicate indexes are allowed but it can often cause confusion, so better to avoid it unless there's a reason, because it returns multiple values in series

```python
import pandas as pd

salaries = [4500, 0, 9500, 1200, 3600, 4520]
departments = ["hr", "developer", "uix designer", "data analyst", "data engineer", "developer"]

salary_series = pd.Series(salaries, index = departments, name = "Salary")
```

```python
salary_series
```

```python
salary_series["hr"]
```

```python
salary_series["developer"]
```
