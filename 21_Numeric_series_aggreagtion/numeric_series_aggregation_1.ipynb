### numeric series aggreagtion

```python
import numpy as np
import pandas as pd

department = ["engineering", "marketing", "sales", "hr", "finance", "support",
"design", "operations", "legal", "research", "admin", "security"]

salary_series = pd.Series([52.0, 47.5, 61.0, np.nan, 58.5, 43.0,
49.5, 54.0, np.nan, 66.0, 41.5, np.nan],
index=department, name="salary")

salary_series
```

```python
# skips NaN
salary_series.count()
```

```python
# does not skip NaN
salary_series.size
```

```python
salary_series.sum()
```

```python
salary_series.mean()
```

```python
salary_series.median()
```

```python
salary_series.min()
```

```python
salary_series.max()
```

```python
salary_series.mean(skipna=False)
```

```python
salary_series.fillna(0).mean()
```

```python
round(salary_series.mean(), 2)
```
