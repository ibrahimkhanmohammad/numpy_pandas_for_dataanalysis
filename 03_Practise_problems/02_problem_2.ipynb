### A teacher has exam scores for 8 students and wants to analyze them with NumPy. Complete the following tasks:



1. Create a 2 x 4 NumPy array called scores using these values:
35, 42, 50, 61, 28, 74, 81, 39.

```python
import numpy as np

scores = np.array([35, 42, 50, 61, 28, 74, 81, 39]).reshape(2, 4)
```

```python
scores
```

2. Add 5 grace marks to every score and store the result in updated_scores.

```python
updated_scores = scores + 5
```

```python
updated_scores
```

3. From updated_scores, extract only the second row.

```python
print(updated_scores[1, :])
```

4. Filter the values in updated_scores that are 50 or higher.

```python
score_mask = updated_scores >= 50
updated_scores[score_mask]
```

5. Use np.where() to create a new array that shows 'Pass' for scores
50 or more and 'Needs Work' for scores below 50.

```python
grade = np.where(updated_scores >= 50, "Pass", "Need work")
```

```python
grade
```
