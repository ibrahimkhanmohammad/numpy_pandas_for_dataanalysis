### 1D array indexing and slicing

```python
# numpy indexing works same as python lists

import numpy as np

cities_array = np.array(["Mumbai", "Hyderabad", "Chennai", "Kolkata", "Pune", "Banglore", "Lucknnow"])

print(cities_array)
print(cities_array[0])
print(cities_array[-1])
print(cities_array[2:5])
print(cities_array[:3])
print(cities_array[1::2])
print(cities_array[::-1])
```
