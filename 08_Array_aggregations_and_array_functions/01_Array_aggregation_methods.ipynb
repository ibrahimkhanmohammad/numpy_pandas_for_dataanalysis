### Array aggregation methods

1D array

```python
import numpy as np

sales = np.array([120, 200, 150, 180, 210, 190, 170])
```

```python
sales
```

```python
# Total sum of sales
sales.sum()
```

```python
# Average of sales
avg_sales = sales.mean()
round(avg_sales, 2)
```

```python
# Max sales
sales.max()
```

```python
# Min sales
sales.min()
```

2D array

```python
import numpy as np

sales = np.array([120, 200, 150, 180, 210, 190, 170, 200]).reshape(2, 4)
```

```python
sales
```

```python
# Total sum of array
sales.sum()
```

```python
# Column wise sum
sales.sum(axis=0)
```

```python
# Row wise sum
sales.sum(axis=1)
```

```python
# Column wise avg
sales.mean(axis=0)
```

```python
# Row wise avg
sales.mean(axis=1)
```

```python
# Max sales per column
sales.max(axis=0)
```

```python
# Max sales per row
sales.max(axis=1)
```

```python
# Min sales per column
sales.min(axis=0)
```

```python
# Min sales per row
sales.min(axis=1)
```

```python
# Sort the sales
sales.sort()
sales
```

Median and percentiles

```python
sales = np.array([120, 200, 150, 180, 210, 190, 170])
```

```python
# Median (middle value after sorting)
np.median(sales)
```

```python
# Percentile 

'''Percentile is calculated using the position of values in sorted data, with interpolation when needed
- 50th percentile = Median
- 25th percentile = First quartile (Q1)
- 75th percentile = Third quartile (Q3)'''

np.percentile(sales, 75)
```
