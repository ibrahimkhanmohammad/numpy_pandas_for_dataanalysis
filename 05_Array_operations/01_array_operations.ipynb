### Array operations

```python
import numpy as np

arr_a = np.array([10, 20, 30, 40, 50])
arr_b = np.array([1, 2, 3, 4, 5])
```

```python
print(arr_a)
print(arr_b)
```

```python
# add a scalar

arr_a + 5
```

```python
# subtract a scalar

arr_a - 5
```

```python
# multiply a scalar

arr_b * 5
```

```python
# divide by a scalar

arr_a / 10
```

```python
# element wise addition

arr_a + arr_b
```

```python
# element wise multiplication

arr_a * arr_b
```
