```python
import numpy as np
import pandas as pd

salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing", "intern", "sales", "support"])

salary_series = pd.Series(salary, index = department, name = "Salary")
```

```python
print(salary_series.loc[salary_series >= 50000])
print('--------')
print(salary_series.loc[salary_series.ge(50000)])
```

```python
print(salary_series.loc[salary_series <= 50000])
print('--------')
print(salary_series.loc[salary_series.le(50000)])
```

```python
print(salary_series.loc[salary_series == 50000])
print('--------')
print(salary_series.loc[salary_series.eq(50000)])
```

```python
print(salary_series.loc[salary_series != 50000])
print('--------')
print(salary_series.loc[salary_series.ne(50000)])
```

```python
print(salary_series.loc[salary_series < 50000])
print('--------')
print(salary_series.loc[salary_series.lt(50000)])
```

```python
print(salary_series.loc[salary_series > 50000])
print('--------')
print(salary_series.loc[salary_series.gt(50000)])
```
