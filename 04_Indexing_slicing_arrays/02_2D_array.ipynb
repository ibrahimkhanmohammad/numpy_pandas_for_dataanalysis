### 2D array indexing and slicing

```python
import numpy as np

cities_array = np.array(["Mumbai", "Hyderabad", "Chennai", "Kolkata",
                         "Pune", "Banglore", "Lucknnow", "Kerala"])

cities_grid = cities_array.reshape(2, 4)
```

```python
print(cities_grid)
```

```python
# accessing only one element

print(cities_grid[1, 3])
```

```python
# accessing block of columns

print(cities_grid[:, 1:])
```

```python
# accessing an entire row

print(cities_grid[0, :])
```

```python
# acessing a column by index

print(cities_grid[:, -1])
```
