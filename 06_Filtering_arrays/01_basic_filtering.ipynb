### Filtering arrays

```python
import numpy as np

ages = np.array([16, 18, 41, 30, 6, 9, 54, 46, 20, 10, 8])
```

```python
adult_mask = ages >= 18
ages[adult_mask]
```

```python
children_mask = ages < 18
ages[children_mask]
```

```python
teen_mask = (ages >= 15) & (ages <= 18)
ages[teen_mask]
```

```python
working_mask = (ages >= 18) & (ages <= 60)
ages[working_mask]
```

```python
special_mask = (ages <= 18) | (ages >= 60)
ages[special_mask]
```
