# Pandas series

A Pandas Series is a simple and powerful structure for storing
one-dimensional labeled data in Python.

1. A Series is a one-dimensional labeled list of data.

2. It stores values together with an index, so each value has a position label.

3. Multiple Series can be combined later to build a Datarrame.

```python
import numpy as np
import pandas as pd
```

```python
salary_values = np.array([2500, 3200, 0, 4500, 9800, 4560])
salary_series = pd.Series(salary_values, name = "salary")
```
