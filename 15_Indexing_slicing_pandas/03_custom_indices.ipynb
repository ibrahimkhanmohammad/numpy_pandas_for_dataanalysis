### Custom indices

```python
import pandas as pd

salaries = [4500, 9500, 2500, 6500, 9000, 6200, 7800, 0]
departments = ["developer", "designer", "engineeer", "doctor", "nurse", "mechanic", "teacher", "tester"]
salaries_series = pd.Series(salaries, index = departments, name = "Salary")
dept_series = pd.Series(departments, name = "Departments")
```

```python
salaries_series
```

```python
salaries_series["doctor"]
```

```python
salaries_series["mechanic"]
```

```python
salaries_series[["designer", "teacher"]]
```

```python
# slicing designer - included unlike in traditional slicing
salaries_series[:"designer"]
```

```python
salaries_series["designer":"mechanic"]
```
