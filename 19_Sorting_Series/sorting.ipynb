### sorting series

```python
import numpy as np
import pandas as pd
```

```python
salaries = np.array([45000, 85000, 60000, 75000, 0, 23000])
departments = np.array(["developer", "marketing", "intern", "sales", "support", "architect"])

salary_series = pd.Series(salaries, index = departments, name = "Salary")
```

```python
salary_series
```

.sort_values() looks for values

```python
salary_series.sort_values()
```

```python
salary_series.sort_values(ascending=False)
```

.sort_index() looks for indexes

```python
salary_series.sort_index()
```

```python
salary_series.sort_index(ascending=False)
```

```python
salary_series
```

```python
salary_series.nlargest(2)
```

```python
salary_series.nsmallest(2)
```

```python
salary_series.rank()
```

```python
salary_series.rank(ascending=False)
```
