### Sorting arrays

```python
import numpy as np

scores = np.array([65, 75, 98, 92, 45, 23, 32, 44, 86])
```

```python
scores
```

```python
# sorting arrays in asc order by using .sort() - eventually it'll completely sort the original array i.e; completely change the original array
scores.sort()
```

```python
scores
```

use np.sort(array) which does not change the original array after sorting

```python
# instead better to use :- np.sort(array)
sorted_scores = np.sort(scores)
```

```python
sorted_scores
```

```python
scores = np.array([65, 75, 98, 92, 45, 23, 32, 44, 86]).reshape(3, 3)
```

```python
scores
```

```python
# row wise sorting
scores.sort(axis=1)
scores
```

```python
# column wise sorting
scores.sort(axis=0)
scores
```
