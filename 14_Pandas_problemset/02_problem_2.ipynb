```python
import numpy as np
import pandas as pd

dates = np.array([
    "2026-06-01", "2026-06-02", "2026-06-03", "2026-06-04", "2026-06-05",
    "2026-06-06", "2026-06-07", "2026-06-08", "2026-06-09", "2026-06-10",
    "2026-06-11", "2026-06-12", "2026-06-13", "2026-06-14", "2026-06-15",
    "2026-06-16", "2026-06-17", "2026-06-18", "2026-06-19", "2026-06-20"
])

stock_price = np.array([
    245.5, 247.0, 246.5, 248.0, 250.0,
    249.5, 251.0, 252.5, 250.5, 253.0,
    254.0, 255.5, 253.5, 256.0, 257.0,
    258.0, 257.5, 259.0, 260.5, 261.0
])

stock_series = pd.Series(stock_price, index=dates, name="Stock Price")

stock_series
```

1. Use iloc to calculate the mean of the first 5 days.

```python
stock_series.iloc[0:6].mean()
```

2. Use iloc to calculate the mean of the last 5 days.

```python
stock_series.iloc[-5:].mean()
```

3. Use loc to calculate the mean from
"2026-06-11" to "2026-06-15".

```python
stock_series.loc["2026-06-11": "2026-06-15"].mean()
```

4. Calculate the maximum stock price between June 5 and June 15, using .loc.

```python
stock_series.loc["2026-06-05": "2026-06-15"].max()
```
