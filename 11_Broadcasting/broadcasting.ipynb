### Broadcasting


Broadcasting allows NumPy to perform operations on arrays with different shapes without manually resizing them.

```python
# 1D array
import numpy as np

a = np.array([10, 20, 30])
b = 5

print(a + b)
```

```python
# 2D array
a = np.array([[10],
              [20],
              [30]])

b = np.array([1, 2, 3, 4])
```

```python
print(a + b)
```
