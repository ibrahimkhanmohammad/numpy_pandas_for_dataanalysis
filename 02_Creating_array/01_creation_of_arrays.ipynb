# Creation of arrays

## by using ones

```python
import numpy as np

ones_array = np.ones((2, 5), dtype=int)
ones_array
```

## by using zeros

```python
import numpy as np

zeros_array = np.zeros((1, 5), dtype=int)
zeros_array
```

## by using arange

```python
import numpy as np

arange_array = np.arange(1, 11, 2)
arange_array
```

## by using linspace

```python
import numpy as np

linspace_array = np.linspace(1, 10, 5, dtype=int)
linspace_array
```

## by using reshape

```python
import numpy as np

# 2 x 3 
temperatures =  [[31, 32, 28],
                [24, 21, 19]]

temp_array = np.array(temperatures)
```

```python
temp_array.reshape(1, 6)
```

```python
temp_array.reshape(3, 2)
```

```python
temp_array.reshape(6, 1)
```
