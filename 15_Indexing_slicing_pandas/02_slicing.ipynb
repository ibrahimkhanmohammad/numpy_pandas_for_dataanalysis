### Slicing

```python
import pandas as pd

salary_series = pd.Series([2500, 4500, 0, 5320, 2100, 7850], name = "Salary")
```

```python
salary_series
```

```python
salary_series[:5]
```

```python
salary_series[1:]
```

```python
salary_series[1:5:2]
```

```python
salary_series[::-1]
```

```python
salary_series[1::2]
```
