### arithmetic operators and methods


```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([52.0, 47.5, 61.0, 44.0, 58.5])
department = np.array(["engineering", "marketing", "sales", "hr","finance"])

salary_series = pd.Series(salary, index=department, name="salary")
salary_series
```

```python
salary_series + 5       # operator
salary_series.add(5)    # method
```

```python
salary_series - 5
salary_series.sub(5)
```

```python
salary_series * 1.5
salary_series.mul(1.5)
```

```python
salary_series / 5
salary_series.div(5)
```

```python
salary_series // 10
salary_series.floordiv(10)
```

```python
salary_series % 5
salary_series.mod(5)
```

```python
salary_series ** 2
salary_series.pow(2)
```
