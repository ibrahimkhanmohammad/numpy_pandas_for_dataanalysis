```python
import numpy as np

temp_array = np.array([29.5, 31.0,
28.7, 32.2, 30.4, 27.9])
```

1. Convert this NumPy array into a Pandas Series
named 'Temperature'.

```python
import pandas as pd
temp_series = pd.Series(temp_array, name = "Temperature")
```

```python
temp_series
```

2. Report the Series name, dtype, size, and index.

```python
# series name
temp_series.name
```

```python
# dtype 
temp_series.dtype
```

```python
# size
temp_series.size
```

```python
# index
temp_series.index
```

3. Find the mean of the values array.

```python
temp_series.mean()
```

4. Convert the Series to an integer data type

```python
temp_series = temp_series.astype(int)
```

5. Recalculate the mean after convertion

```python
temp_series.mean()
```
