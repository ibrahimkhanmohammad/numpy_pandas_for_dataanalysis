### A teacher has 8 quiz scores and wants to practice NumPy on them. Complete the following tasks.

1. Create a 1D NumPy array named "scores" using:
16, 25, 36, 25, 49, 64, 81, 100.

```python
import numpy as np

scores = np.array([16, 25, 36, 25, 49, 64, 81, 100])
```

```python
scores
```

2. Update every score below 25 to 25.

```python
scores = np.where(scores < 25, 25, scores)
```

```python
scores
```

3. Create a new array named "sorted_scores" by sorting the updated scores
in ascending order.

```python
sorted_scores = np.sort(scores)
```

```python
sorted_scores
```

4. From the updated array, find the total, average, highest value, and minumum value.

```python
# total 
sorted_scores.sum()
```

```python
# average
np.median(sorted_scores)
```

```python
# highest value
sorted_scores.max()
```

```python
# minimum value
sorted_scores.min()
```

5. Use "np.unique()" to get the distinct score values.

```python
unique_scores = np.unique(sorted_scores)
```

```python
unique_scores
```

6. Use "np.sqrt()" to calculate the square root of each updated score.

```python
sqrt_scores = np.sqrt(sorted_scores)
```

```python
sqrt_scores
```
