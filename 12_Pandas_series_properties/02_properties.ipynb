### Pandas series properties

```python
import numpy as np
import pandas as pd
```

```python
salary_values = np.array([2500, 3200, 0, 4500, 9800, 4560])
salary_series = pd.Series(salary_values, name = "salary")
```

```python
# values - gives you array values
salary_series.values
```

```python
# index - gives you the index of an array
salary_series.index
```

```python
# dtype - gives you data type of ele's of an array
salary_series.dtype
```

```python
# name - gives you the name you provided when creating the series variable
salary_series.name
```
