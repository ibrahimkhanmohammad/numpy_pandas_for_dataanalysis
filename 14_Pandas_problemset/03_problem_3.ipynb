```python
import numpy as np
import pandas as pd

dates = np. array([
"2026-06-01","2026-06-02","2026-06-03","2026-06-04",
"2026-06-05","2026-06-06", "2026-06-07","2026-06-08",
"2026-06-09", "2026-06-10", "2026-06-11","2026-06-12",
"2026-06-13", "2026-06-14", "2026-06-15", "2026-06-16",
"2026-06-17", "2026-06-18", "2026-06-19","2026-06-20"])

stock_price = np.array([
124.5, 126.2, 123.8, 121.9, 125.4,
127.1, 122.6, 120.8, 119.7, 121.3,
118.9, 120.2, 122.1, 123.4, 119.8,
117.6, 118.4, 120.1, 121.7, 122.9])

stock_series = pd.Series(stock_price, index=dates, name="Stock Price")
```

1. Find the 10 lowest stock prices.

```python
updated_stock_series = stock_series.nsmallest(10)
updated_stock_series
```

2. Display those 10 rows starting with the most recent date.

```python
updated_stock_series.sort_index(ascending=False, inplace=True)
updated_stock_series
```

3. Show only the stock prices less than or equal to 120.5.

```python
updated_stock_series.loc[updated_stock_series <= 120.5]
```
