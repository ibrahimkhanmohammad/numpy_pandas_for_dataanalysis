### Filtering an array based on another array

```python
import numpy as np

ages = np.array([18, 21, 14, 32, 40, 46, 12, 8])
names = np.array(['Rahul', 'Priya', 'Arjun', 'Sneha', 'Vikram', 'Ananya', 'Kiran', 'Aisha'])
```

```python
# print names of people who are adult i.e; whose age is >= 18

adult_mask = ages >= 18
ages[adult_mask]
names[adult_mask]

# adult_mask will work with names, because length of both ages and names are of same
```
