### loc method

loc lets you access databy labels. You use the names in the index, not positions.

```python
import pandas as pd

salaries = [2500, 0, 9500, 4600, 8500, 3250]
role = ["designer", "engineer", "developer", "doctor", "chef", "mechanic"]

salaries_series = pd.Series(salaries, index = role, name = "Salaries")
role_series = pd.Series(role, name = "Role")
```

```python
salaries_series
```

```python
salaries_series.loc["engineer"]
```

```python
salaries_series.loc["doctor":"mechanic"]
```

```python
salaries_series.loc[["designer", "doctor"]]
```

```python
salaries_series.loc[:"mechanic"]
```

```python
salaries_series.loc["engineer"] = 6500
salaries_series.loc["engineer"]
```

```python
salaries_series
```

```python
salaries_series.loc[salaries_series >= 4500] 
```

```python
salaries_series.loc[salaries_series == 0] 
```
