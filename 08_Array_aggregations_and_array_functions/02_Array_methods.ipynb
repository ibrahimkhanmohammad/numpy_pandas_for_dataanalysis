### Array Functions

```python
# Array functions 

import numpy as np

sales = np.array([150, 200, 150, 300, 250, 200, 350, 150]) # has duplicates like 150 and 200
```

```python
sales
```

```python
# Unique (distinct elements of an array)
np.unique(sales)
```

```python
# return_counts = True with unique returns how many times each element repeated
np.unique(sales, return_counts=True)
```

```python
# Square root of each value of an array
np.sqrt(sales)
```

```python
# Std deviation of an array
np.std(sales)
```

```python
# Cummulative sum of an array - it prints the sum of all values at that point
np.cumsum(sales)
```

```python
# Round each value to 2 decimal places
np.round(np.sqrt(sales), 2)
```

```python
# Absolute value - it returns the absolute value of an element inside an array such that +ve ele remains same and -ve ele remains -ve
changes = np.array([-20, 25, 15, 55, -94])
np.abs(changes)
```
